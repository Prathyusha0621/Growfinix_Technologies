```python
df = pd.read_csv("Growfinix_Task3_Tour_Enquiries.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
destination_counts = df["Destination"].value_counts()

destination_counts
```

```python
import matplotlib.pyplot as plt

destination_counts.plot(kind="bar", figsize=(10, 5))

plt.title("Most Popular Travel Destinations")
plt.xlabel("Destination")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
bins = [17, 25, 35, 45, 55, 65]
labels = ["18-25", "26-35", "36-45", "46-55", "56-65"]

df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=labels)

age_group_counts = df["Age_Group"].value_counts().sort_index()

age_group_counts
```

```python
age_group_counts.plot(kind="bar", figsize=(8, 5))

plt.title("Customers by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
income_counts = df["Income_Bracket"].value_counts()

income_counts
```

```python
income_counts.plot(kind="bar", figsize=(8, 5))

plt.title("Customers by Income Bracket")
plt.xlabel("Income Bracket")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df.groupby("Income_Bracket")["Income"].mean()
```

```python
df.groupby("Income_Bracket")["Income"].mean().plot(kind="bar", figsize=(8, 5))

plt.title("Average Income by Income Bracket")
plt.xlabel("Income Bracket")
plt.ylabel("Average Income")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
